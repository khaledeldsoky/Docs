#### 1 - Scale a Deployment

##### Scale an existing deployment to a specific number of replicas ( Doc ✅)


```python
kubectl scale deployment <deployment-name> --replicas=<number>
```

#### 2 - Edit Deployment & Expose via NodePort

##### Add ports to a deployment manifest without a local file and expose it through a service

```python
kubectl edit deployment <deployment-name> # (to add container ports manually)

kubectl expose deployment <deployment-name> --type=NodePort --port=<service-port> --target-port=<container-port>

kubectl get svc # (to verify the NodePort value)
```

#### 3 - Persistent Volume Claim (PVC) & Pod Mounting

##### Create a 4GB ReadWriteOnce PVC and mount it to a pod

```python
kubectl apply -f pvc.yaml.
kubectl apply -f pod.yaml.
kubectl describe pod <pod-name> # (to check for "Unbound" status or successful mounting)
```

#### 4 - Create a Persistent Volume (PV)

##### Create a 6GB ReadOnlyMany PV using a hostPath

```python
kubectl apply -f pv.yaml.
kubectl get pv
```

#### 5 -Configure Ingress

##### Create an Ingress named "luau" routing /aloha to the "Aloha" service

```python
kubectl apply -f ingress.yaml
kubectl describe ingress luau
```

#### 6 - Identify High CPU Usage

##### Find the pod with label app=intensive in the integration namespace using the most CPU and save its name to a file

```python
kubectl top pods -n integration
kubectl get pods -n integration -L app # (to filter by labels)
echo <pod-name> > <file-path>
```

#### 7 - Schedule Pod via NodeSelector

##### Force a pod onto a node labeled disk=nvme

```python
kubectl get nodes -L disk # (to find the matching node).
kubectl apply -f pod.yaml # (containing the nodeSelector).
kubectl get pod <pod-name> -o wide # (to confirm the node assignment)
```

#### 14 - Secrets as Environment Variables

##### Extract an existing pod's YAML, create a Secret, and update the pod to use Secret values as env vars

```python
kubectl get pod <name> -n kiwi -o yaml > pod.yaml
kubectl apply -f secret.yaml
kubectl delete pod <name> -n kiwi
kubectl apply -f pod.yaml
kubectl exec -it <pod-name> -n kiwi -- bash 
```

#### 19 - Cluster Upgrade

##### Upgrade a controller node to version 1.32.1

```python
apt-mark unhold kubeadm, apt-get update
apt-get install -y kubeadm=1.32.1-1.1
kubectl drain <node> --ignore-daemonsets
kubeadm upgrade plan  
sudo kubeadm upgrade apply v1.32.1
apt-get install -y kubelet=1.32.1-1.1 kubectl=1.32.1-1.1
systemctl daemon-reload  
systemctl restart kubelet
```

#### 20 - ETCD Backup & Restore

##### Save an etcd snapshot and restore the cluster state

```python
ETCDCTL_API=3 etcdctl snapshot save <file> --cacert=<path> --cert=<path> --key=<path>
ETCDCTL_API=3 etcdctl snapshot restore <file> --data-dir=<new-dir>
vi /etc/kubernetes/manifests/etcd.yaml # (to update the hostPath for the data volume)
```
